```python

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

# Load data
file_path = 'Health_Screening_Labeled_Numeric.csv'
data = pd.read_csv(file_path)

columns_to_exclude = ['id', 'gender','height','weight','Sistolik','Diastolik','Age','BMI']
features = data.drop(columns=['Health_Status'] + columns_to_exclude)
target = data['Health_Status']

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.01,
    random_state=42
)
```

```python

results = []

from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(max_iter=1000)
logreg.fit(X_train, y_train)
y_pred = logreg.predict(X_test)

# Calculate metrics
logreg_metrics = {
    'Model': 'Logistic Regression',
    'Accuracy': accuracy_score(y_test, y_pred),
    'Precision': precision_score(y_test, y_pred, average='weighted'),
    'Recall': recall_score(y_test, y_pred, average='weighted'),
    'F1': f1_score(y_test, y_pred, average='weighted')
}
results.append(logreg_metrics)

print("Logistic Regression Report:")
print(classification_report(y_test, y_pred))
```

```python
# ---------------------------
# 6. Model 3: Random Forest
# ---------------------------
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)

rf_metrics = {
    'Model': 'Random Forest',
    'Accuracy': accuracy_score(y_test, y_pred),
    'Precision': precision_score(y_test, y_pred, average='weighted'),
    'Recall': recall_score(y_test, y_pred, average='weighted'),
    'F1': f1_score(y_test, y_pred, average='weighted')
}
results.append(rf_metrics)

print("\nRandom Forest Report:")
print(classification_report(y_test, y_pred))
```

```python
# ---------------------------
# 7. Model 4: Gradient Boosting
# ---------------------------
from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier()
gb.fit(X_train, y_train)
y_pred = gb.predict(X_test)

gb_metrics = {
    'Model': 'Gradient Boosting',
    'Accuracy': accuracy_score(y_test, y_pred),
    'Precision': precision_score(y_test, y_pred, average='weighted'),
    'Recall': recall_score(y_test, y_pred, average='weighted'),
    'F1': f1_score(y_test, y_pred, average='weighted')
}
results.append(gb_metrics)

print("\nGradient Boosting Report:")
print(classification_report(y_test, y_pred))
```

```python
# ---------------------------
# 8. Model 5: XGBoost
# ---------------------------
from xgboost import XGBClassifier

xgb = XGBClassifier()
xgb.fit(X_train, y_train)
y_pred = xgb.predict(X_test)

xgb_metrics = {
    'Model': 'XGBoost',
    'Accuracy': accuracy_score(y_test, y_pred),
    'Precision': precision_score(y_test, y_pred, average='weighted'),
    'Recall': recall_score(y_test, y_pred, average='weighted'),
    'F1': f1_score(y_test, y_pred, average='weighted')
}
results.append(xgb_metrics)

print("\nXGBoost Report:")
print(classification_report(y_test, y_pred))
```

```python
# ---------------------------
# 11. Model 6: K-Nearest Neighbors (KNN)
# ---------------------------
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

knn_metrics = {
    'Model': 'KNN',
    'Accuracy': accuracy_score(y_test, y_pred),
    'Precision': precision_score(y_test, y_pred, average='weighted'),
    'Recall': recall_score(y_test, y_pred, average='weighted'),
    'F1': f1_score(y_test, y_pred, average='weighted')
}
results.append(knn_metrics)

print("\nKNN Report:")
print(classification_report(y_test, y_pred))
```

```python
# ---------------------------
# 12. Model 7: Naive Bayes
# ---------------------------
from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()
nb.fit(X_train, y_train)
y_pred = nb.predict(X_test)

nb_metrics = {
    'Model': 'Naive Bayes',
    'Accuracy': accuracy_score(y_test, y_pred),
    'Precision': precision_score(y_test, y_pred, average='weighted'),
    'Recall': recall_score(y_test, y_pred, average='weighted'),
    'F1': f1_score(y_test, y_pred, average='weighted')
}
results.append(nb_metrics)

print("\nNaive Bayes Report:")
print(classification_report(y_test, y_pred))
```

```python
# ---------------------------
# 13. Model 8: Artificial Neural Network (ANN)
# ---------------------------
from sklearn.neural_network import MLPClassifier

ann = MLPClassifier(
    hidden_layer_sizes=(100, 50),
    activation='relu',
    solver='adam',
    max_iter=1000,
    random_state=42
)
ann.fit(X_train, y_train)
y_pred = ann.predict(X_test)

ann_metrics = {
    'Model': 'ANN',
    'Accuracy': accuracy_score(y_test, y_pred),
    'Precision': precision_score(y_test, y_pred, average='weighted'),
    'Recall': recall_score(y_test, y_pred, average='weighted'),
    'F1': f1_score(y_test, y_pred, average='weighted')
}
results.append(ann_metrics)

print("\nANN Report:")
print(classification_report(y_test, y_pred))
```

```python
# ---------------------------
# 1. Import Libraries & Load Data
# ---------------------------
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

# Load data
file_path = 'Health_Screening_Labeled_Numeric_v2.csv'
data = pd.read_csv(file_path)

# ---------------------------
# 2. Preprocessing Data
# ---------------------------
columns_to_exclude = ['id', 'gender','height','weight','Sistolik','Diastolik','Age','BMI']
features = data.drop(columns=['Health_Status'] + columns_to_exclude)
target = data['Health_Status']

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.2,
    random_state=42
)
```

```python
# ---------------------------
# 5. Model 2: Support Vector Machine
# ---------------------------
from sklearn.svm import SVC

svm = SVC()
svm.fit(X_train, y_train)
y_pred = svm.predict(X_test)

svm_metrics = {
    'Model': 'SVM',
    'Accuracy': accuracy_score(y_test, y_pred),
    'Precision': precision_score(y_test, y_pred, average='weighted'),
    'Recall': recall_score(y_test, y_pred, average='weighted'),
    'F1': f1_score(y_test, y_pred, average='weighted')
}
results.append(svm_metrics)

print("\nSVM Report:")
print(classification_report(y_test, y_pred))
```
