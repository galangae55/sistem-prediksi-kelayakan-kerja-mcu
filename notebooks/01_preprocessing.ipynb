```python
import numpy as np
import pandas as pd

file_path = 'BMI_Health_Screening_Data_Expanded_v2.csv'
data = pd.read_csv(file_path)

num_samples = len(data)

normal_percentage = 0.70
num_normal = int(num_samples * normal_percentage)
num_abnormal = num_samples - num_normal

# Generate normal data
bmi_min_normal = 18.5
bmi_max_normal = 22.9

height_normal = np.random.uniform(150, 180, num_normal).astype(int)
weight_normal = (np.random.uniform(bmi_min_normal, bmi_max_normal) * (height_normal/100)**2).astype(int)

normal_sistolik_range = (90, 120)
normal_diastolik_range = (60, 80)
sistolik_normal = np.random.randint(*normal_sistolik_range, num_normal)
diastolik_normal = np.random.randint(*normal_diastolik_range, num_normal)

# Generate abnormal data
height_abnormal = np.random.randint(120, 200, num_abnormal)
weight_abnormal = np.random.randint(30, 120, num_abnormal)
sistolik_abnormal = np.random.randint(80, 180, num_abnormal)
diastolik_abnormal = np.random.randint(50, 110, num_abnormal)

# Gabungkan dan acak data
def combine_and_shuffle(normal_data, abnormal_data):
    combined = np.concatenate([normal_data, abnormal_data])
    np.random.shuffle(combined)  # Acak urutan data
    return combined

height = combine_and_shuffle(height_normal, height_abnormal)
weight = combine_and_shuffle(weight_normal, weight_abnormal)
sistolik = combine_and_shuffle(sistolik_normal, sistolik_abnormal)
diastolik = combine_and_shuffle(diastolik_normal, diastolik_abnormal)

# Update DataFrame dengan urutan teracak
data['height'] = height
data['weight'] = weight
data['Sistolik'] = sistolik
data['Diastolik'] = diastolik

# Hitung BMI setelah pengacakan
data['BMI'] = (data['weight'] / ((data['height']/100)**2)).astype(int)

output_path = 'BMI_Health_Screening_Data_Updated.csv'
data.to_csv(output_path, index=False)

print(f"Data updated and saved to {output_path}")
```

```python
import pandas as pd

file_path = 'BMI_Health_Screening_Data_Updated.csv'
data = pd.read_csv(file_path)
def classify_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi <= 22.9:
        return "Normal"
    elif 23 <= bmi <= 24.9:
        return "Overweight"
    elif 25 <= bmi <= 29.9:
        return "Obesitas I"
    else: 
        return "Obesitas II"


data['BMICat'] = data['BMI'].apply(classify_bmi)

output_file_path = 'BMI_Health_Screening_Data.csv'
data.to_csv(output_file_path, index=False)

print(f"Dataset dengan kategori BMI baru telah disimpan ke: {output_file_path}")
```

```python

def hipertensi_kategori(row):
    sistolik = row['Sistolik']
    diastolik = row['Diastolik']

    if sistolik < 120 and diastolik < 80:
        return 'Normal'

    elif 120 <= sistolik <= 139 or 80 <= diastolik <= 89:
        return 'Pra-hipertensi'
    elif 140 <= sistolik <= 159 or 90 <= diastolik <= 99:
        return 'Hipertensi Tingkat 1'

    elif sistolik >= 160 or diastolik >= 100:
        return 'Hipertensi Tingkat 2'
    elif sistolik > 140 and diastolik < 90:
        return 'Tidak Diketahui'

data['Hipertensi_Kategori'] = data.apply(hipertensi_kategori, axis=1)

print("\nData setelah kolom diubah:")
print(data.head())

output_file = 'BMI_Health_Screening_Data_Updated.csv'
data.to_csv(output_file, index=False)
print(f"Data dengan kolom baru telah disimpan ke {output_file}")
```
