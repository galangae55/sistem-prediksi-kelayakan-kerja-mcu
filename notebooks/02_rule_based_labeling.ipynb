```python
import pandas as pd

def evaluate_health_impact(row):
    
    severe_conditions = [
        # Kelainan organ vital
        row['Jantung'] == 'Tidak normal', 
        # Gangguan sistemik berat
        row['gangguan jiwa'] == 'ada',
        row['trauma pada kepala'] == 'ada',
        row['hepatitis'] == 'ada',
        
        # Kardiovaskular
        row['Treadmil'] == 'positif',
    ]

    restricted_conditions = [
        row['Hipertensi_Kategori'] == 'Hipertensi Tingkat 2',
        row['BMICat'] == 'Obesitas I',
        row['BMICat'] == 'Obesitas II',   
        # Gangguan sensorik
        row['Buta Warna'] == 'ya',
        row['Audiometri'] not in ['normal'],
        row['Spirometri'] == 'tidak normal',
        
        # Pernapasan abnormal
        row['Frekuensi Napas (kali/menit)'] == 'Tidak normal',
        
        # Penyakit kronis
        row['diabetes melitus'] == 'ada',
        row['hipertensi'] == 'ada',
        row['penyakit jantung'] == 'ada',
        row['tuberkulosis'] == 'ada',
        
        # Parameter darah kritis
        row['gula darah puasa'] == 'diatas 200',
        row['gula darah 2 jam pp'] == 'diatas 200',
        row['B U N'] == 'tidak normal',
        row['kreatinin'] == 'tidak normal',
        row['reduksi pada urine'] == 'positif 2',
        row['HBs Ag Kuantitatif'] == 'positif'
    ]

    note_conditions = [
        row['Hipertensi_Kategori'] == 'Pra-hipertensi',  
        # Kriteria gula darah baru
        row['gula darah puasa'] == 'tidak normal',
        row['gula darah 2 jam pp'] == 'tidak normal',
        # Parameter darah abnormal
        row['hemoglobin'] == 'tidak normal',
        row['leukosit'] == 'tidak normal',
        row['eritrosit'] == 'tidak normal',
        row['hematokrit'] == 'tidak normal',
        row['trombosit'] == 'tidak normal',
        row['LED'] == 'tidak normal',
        row['eosinofil'] == 'tidak normal',
        row['basofil'] == 'tidak normal',
        row['neutrofil'] == 'tidak normal',
        row['lymphosit'] == 'tidak normal',
        row['monosit'] == 'tidak normal',
        row['MCV'] == 'tidak normal',
        row['S G O T'] == 'tidak normal',
        row['S G P T'] == 'tidak normal',
        row['anti HBs'] == 'negatif',
        
        # Parameter biokimia
        row['asam urat'] == 'tidak normal',
        row['kolestrol total'] == 'tidak normal',
        row['trigliserida'] == 'tidak normal',
        row['kolestrol HDL (direct)'] == 'tidak normal',
        row['kolestrol LDL (direct)'] == 'tidak normal',
        
        # Parameter urine lengkap
        row['pH pada urine'] == 'tidak normal',
        row['berat jenis pada urine'] == 'tidak normal',
        row['eritrosit pada urine'] == 'tidak normal',
        row['lekosit pada urine'] == 'tidak normal',
        row['nitrite pada urine'] == 'positif',
        row['protein pada urine'] == 'positif',
        row['reduksi pada urine'] == 'positif 1',
        row['ketone pada urine'] == 'positif',
        row['urobilinogen pada urine'] == 'positif',
        row['billirubin pada urine'] == 'positif',
        row['silinder pada urine'] == 'positif',
        row['kristal pada urine'] == 'positif',
        row['yeast pada urine'] == 'positif',
        row['bakteri pada urine'] == 'positif',
        
        # Parameter tambahan
        row['Tingkatan Kesadaran'] not in ['compos mentis'],
        row['BMICat'] == 'Underweight',
        row['BMICat'] == 'Overweight',
        row['Nadi (kali/menit)'] == 'Tidak normal',
        row['foto thorax'] == 'tidak normal',
        row['Hipertensi_Kategori'] == 'Hipertensi Tingkat 1',
    ]

    # Evaluasi akhir
    if any(severe_conditions):
        return "UNFIT"
    elif any(restricted_conditions):
        return "FIT WITH RESTRICTION"
    elif any(note_conditions):
        return "FIT WITH NOTE"
    else:
        return "FIT"

# Eksekusi utama
if __name__ == "__main__":
    data = pd.read_csv("updated_health_screening_data.csv")
    
    data['Health_Status'] = data.apply(evaluate_health_impact, axis=1)
    data.to_csv("Health_Screening_Fully_Labeled_v3.csv", index=False)
    print("Proses labeling selesai dengan semua parameter terintegrasi!")
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

file_path = 'Health_Screening_Fully_Labeled_v3.csv'
data = pd.read_csv(file_path)


non_numeric_columns = data.select_dtypes(include=['object']).columns

label_encoders = {}


for column in non_numeric_columns:
    le = LabelEncoder()
    data[column] = le.fit_transform(data[column])
    label_encoders[column] = le


data.to_csv('Health_Screening_Labeled_Numeric_v2.csv', index=False)


print(f"Converted non-numeric columns: {list(non_numeric_columns)}")
print("Dataset has been saved as 'Health_Screening_Labeled_Numeric.csv'.")


print(data[non_numeric_columns].head())
```

```python

for column, le in label_encoders.items():
    print(f"\n### {column}")
    for class_, value in zip(le.classes_, le.transform(le.classes_)):
        print(f"- `{class_}` -> {value}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
import joblib
import pandas as pd

file_path = 'Health_Screening_Labeled_Numeric.csv'
data = pd.read_csv(file_path)


columns_to_exclude = ['id', 'gender','height','weight','Sistolik','Diastolik','Age','BMI']  
features = data.drop(columns=['Health_Status'] + columns_to_exclude)
target = data['Health_Status']


X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.2,
    random_state=42
)

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("\nModel Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print(classification_report(y_test, y_pred))

```
