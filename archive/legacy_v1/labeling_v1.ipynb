```python
import pandas as pd

def evaluate_health_impact(row):
    # Konversi nilai gender ke integer jika diperlukan
    gender = int(row['gender']) if isinstance(row['gender'], str) else row['gender']
    
    severe_conditions = [
        # Kelainan organ vital
        row['Jantung'] == 'Tidak normal',
        row['Spirometri'] == 'tidak normal',
        
        
        # Gangguan sistemik berat
        row['gangguan jiwa'] == 'ada',
        row['trauma pada kepala'] == 'ada',
        row['hepatitis'] == 'ada',
        row['HBs Ag Kuantitatif'] == 'positif',
        
        # Kardiovaskular
        row['Treadmil'] == 'positif',
        row['BMICat'] in ['Obesitas II']
    ]

    restricted_conditions = [

        row['BMICat'] in ['Obesitas I'],
        # Gangguan sensorik
        row['Buta Warna'] == 'ya',
        row['Audiometri'] not in ['normal'],
        
        # Pernapasan abnormal
        (row['Frekuensi Napas (kali/menit)'] < 12) | (row['Frekuensi Napas (kali/menit)'] > 20),
        
        # Penyakit kronis
        row['diabetes melitus'] == 'ada',
        row['hipertensi'] == 'ada',
        row['penyakit jantung'] == 'ada',
        row['tuberkulosis'] == 'ada',
        
        # Parameter darah kritis
        (row['gula darah puasa'] > 200) | (row['gula darah 2 jam pp'] > 200),
        (row['B U N'] < 5) | (row['B U N'] > 23.5),
        (row['kreatinin'] < (0.7 if gender == 2 else 0.8)) | (row['kreatinin'] > (1.1 if gender == 2 else 1.4)),
        row['reduksi pada urine'] == 'positif 2'
    ]

    note_conditions = [

        # Kriteria gula darah baru
        (row['gula darah puasa'] >= 126) & (row['gula darah puasa'] <= 200),
        (row['gula darah 2 jam pp'] >= 140) & (row['gula darah 2 jam pp'] <= 200),
        # Parameter darah abnormal
        (row['hemoglobin'] < (11.5 if gender == 2 else 13)) | (row['hemoglobin'] > (16.5 if gender == 2 else 18)),
        (row['leukosit'] < 4000) | (row['leukosit'] > 11000),
        (row['eritrosit'] < (4 if gender == 2 else 4.5)) | (row['eritrosit'] > (5 if gender == 2 else 5.5)),
        (row['hematokrit'] < (37 if gender == 2 else 40)) | (row['hematokrit'] > (45 if gender == 2 else 50)),
        (row['trombosit'] < 150000) | (row['trombosit'] > 400000),
        (row['LED'] < (0 if gender == 2 else 0)) | (row['LED'] > (20 if gender == 2 else 15)),
        row['eosinofil'] > 1,
        row['basofil'] > 1,
        (row['neutrofil'] < 50) | (row['neutrofil'] > 70),
        (row['lymphosit'] < 25) | (row['lymphosit'] > 33),
        (row['monosit'] < 3) | (row['monosit'] > 8),
        (row['MCV'] < 82) | (row['MCV'] > 92),
        row['S G O T'] > (31 if gender == 2 else 37),
        row['S G P T'] > (32 if gender == 2 else 42),
        
        # Parameter biokimia
        (row['asam urat'] < (2.6 if gender == 2 else 3.5)) | (row['asam urat'] > (6 if gender == 2 else 7.2)),
        row['kolestrol total'] > 200,
        row['trigliserida'] > 200,
        row['kolestrol HDL (direct)'] < 35,
        row['kolestrol LDL (direct)'] > 115,
        
        # Parameter urine lengkap
        (row['pH pada urine'] < 5) | (row['pH pada urine'] > 8),
        (row['berat jenis pada urine'] < 1.005) | (row['berat jenis pada urine'] > 1.030),
        row['eritrosit pada urine'] > 2,
        row['lekosit pada urine'] > 5,
        row['nitrite pada urine'] == 'positif',
        row['protein pada urine'] == 'positif',
        row['reduksi pada urine'] == 'positif 1',
        row['ketone pada urine'] == 'positif',
        row['urobilinogen pada urine'] == 'positif',
        row['billirubin pada urine'] == 'positif',
        row['silinder pada urine'] == 'positif',
        row['kristal pada urine'] == 'positif',
        row['yeast pada urine'] == 'positif',
        row['bakteri pada urine'] == 'positif',
        
        # Parameter tambahan
        row['Tingkatan Kesadaran'] not in ['compos mentis'],
        row['BMICat'] in ['Underweight', 'Overweight',],
        (row['Nadi (kali/menit)'] < 60) | (row['Nadi (kali/menit)'] > 100),
        row['foto thorax'] == 'tidak normal',
    ]

    # Evaluasi tekanan darah dengan logika lebih akurat
    sistolik = row['Sistolik']
    diastolik = row['Diastolik']
    
    if sistolik >= 160 or diastolik >= 100:  
        restricted_conditions.append(True)
    elif sistolik > 140 and diastolik < 90:  
        restricted_conditions.append(True)
    elif 140 <= sistolik <= 159 or 90 <= diastolik <= 99: 
        note_conditions.append(True)
    elif 120 <= sistolik <= 139 or 80 <= diastolik <= 89:  
        note_conditions.append(True) 

    # Evaluasi akhir
    if any(severe_conditions):
        return "UNFIT"
    elif any(restricted_conditions):
        return "FIT WITH RESTRICTION"
    elif any(note_conditions):
        return "FIT WITH NOTE"
    else:
        return "FIT"

# Eksekusi utama
if __name__ == "__main__":
    data = pd.read_csv("dataset_berat_jenis_updated.csv")
    
    # Handle missing values dan konversi tipe data
    numeric_cols = ['hemoglobin', 'kreatinin', 'asam urat']
    data[numeric_cols] = data[numeric_cols].apply(pd.to_numeric, errors='coerce')
    data.fillna({col: data[col].median() for col in numeric_cols}, inplace=True)
    
    data['Health_Status'] = data.apply(evaluate_health_impact, axis=1)
    data.to_csv("Health_Screening_Fully_Labeled_v2.csv", index=False)
    print("Proses labeling selesai dengan semua parameter terintegrasi!")
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

file_path = 'Health_Screening_Fully_Labeled_v2.csv'
data = pd.read_csv(file_path)


non_numeric_columns = data.select_dtypes(include=['object']).columns

label_encoders = {}


for column in non_numeric_columns:
    le = LabelEncoder()
    data[column] = le.fit_transform(data[column])
    label_encoders[column] = le


data.to_csv('Health_Screening_Labeled_Numeric.csv', index=False)


print(f"Converted non-numeric columns: {list(non_numeric_columns)}")
print("Dataset has been saved as 'Health_Screening_Labeled_Numeric.csv'.")


print(data[non_numeric_columns].head())
```

```python

for column, le in label_encoders.items():
    print(f"\n### {column}")
    for class_, value in zip(le.classes_, le.transform(le.classes_)):
        print(f"- `{class_}` -> {value}")
```

```python
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, accuracy_score
import joblib
import time  # Untuk mengukur waktu training

# Load data
X = data.drop(columns=['Health_Status']) 
y = data['Health_Status']

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y, 
    test_size=0.1, 
    random_state=42,
    stratify=y  # Penting untuk data tidak seimbang
)

# Inisialisasi model XGBoost dengan parameter optimal untuk data besar
model = XGBClassifier(
    n_estimators=200,        # Jumlah pohon lebih banyak untuk data besar
    max_depth=8,             # Kedalaman lebih untuk menangkap pola kompleks
    learning_rate=0.1,       # Learning rate lebih rendah untuk stabilitas
    subsample=0.8,           # Gunakan 80% data tiap iterasi untuk prevent overfit
    colsample_bytree=0.8,    # Gunakan 80% fitur tiap pohon
    objective='binary:logistic',  # Sesuaikan dengan jenis masalah klasifikasi Anda
    random_state=42,
    n_jobs=-1,               # Gunakan semua core CPU
    tree_method='hist'       # Metode optimal untuk data besar
)

# Training dengan timer
start_time = time.time()
model.fit(X_train, y_train)
print(f"Training time: {time.time() - start_time:.2f} detik")

# Prediksi & evaluasi
y_pred = model.predict(X_test)

print("\nEvaluasi Model:")
print(f"Akurasi: {accuracy_score(y_test, y_pred):.2f}")
print(classification_report(y_test, y_pred))

# Simpan model
model_path = 'health_status_model_xgboost_v1.pkl'
joblib.dump(model, model_path)
print(f"Model telah disimpan di {model_path}")
```
