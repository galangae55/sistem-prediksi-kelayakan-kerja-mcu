```python
import numpy as np
import pandas as pd


file_path = 'BMI_Health_Screening_Data_Expanded.csv'
data = pd.read_csv(file_path)


num_samples = len(data)


normal_percentage = 0.90
num_normal = int(num_samples * normal_percentage)
num_abnormal = num_samples - num_normal


bmi_min = 18.5
bmi_max = 22.9


height_normal = np.random.uniform(150, 180, num_normal).astype(int)
weight_normal = [(np.random.uniform(bmi_min, bmi_max) * (h / 100) ** 2) for h in height_normal]
weight_normal = np.array(weight_normal).astype(int)  


height_abnormal = np.random.uniform(120, 200, num_abnormal).astype(int)  
weight_abnormal = np.random.uniform(30, 120, num_abnormal).astype(int)  

normal_sistolik_range = (90, 120)  
normal_diastolik_range = (60, 80)  
sistolik_normal = np.random.uniform(normal_sistolik_range[0], normal_sistolik_range[1], num_normal).astype(int)
diastolik_normal = np.random.uniform(normal_diastolik_range[0], normal_diastolik_range[1], num_normal).astype(int)

sistolik_abnormal = np.random.uniform(80, 180, num_abnormal).astype(int)  
diastolik_abnormal = np.random.uniform(50, 110, num_abnormal).astype(int)  

height = np.concatenate((height_normal, height_abnormal))
weight = np.concatenate((weight_normal, weight_abnormal))
sistolik = np.concatenate((sistolik_normal, sistolik_abnormal))
diastolik = np.concatenate((diastolik_normal, diastolik_abnormal))


data['height'] = height
data['weight'] = weight
data['Sistolik'] = sistolik
data['Diastolik'] = diastolik


data['BMI'] = (data['weight'] / ((data['height'] / 100) ** 2)).astype(int)


output_path = 'BMI_Health_Screening_Data_Updated.csv'
data.to_csv(output_path, index=False)

print(f"Data updated and saved to {output_path}")
```

```python
import pandas as pd

file_path = 'BMI_Health_Screening_Data_Updated.csv'
data = pd.read_csv(file_path)
def classify_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi <= 22.9:
        return "Normal"
    elif 23 <= bmi <= 24.9:
        return "Overweight"
    elif 25 <= bmi <= 29.9:
        return "Obesitas I"
    else: 
        return "Obesitas II"


data['BMICat'] = data['BMI'].apply(classify_bmi)

output_file_path = 'BMI_Health_Screening_Data.csv'
data.to_csv(output_file_path, index=False)

print(f"Dataset dengan kategori BMI baru telah disimpan ke: {output_file_path}")
```

```python

def hipertensi_kategori(row):
    sistolik = row['Sistolik']
    diastolik = row['Diastolik']

    if sistolik < 120 and diastolik < 80:
        return 'Normal'

    elif 120 <= sistolik <= 139 or 80 <= diastolik <= 89:
        return 'Pra-hipertensi'
    elif 140 <= sistolik <= 159 or 90 <= diastolik <= 99:
        return 'Hipertensi Tingkat 1'

    elif sistolik >= 160 or diastolik >= 100:
        return 'Hipertensi Tingkat 2'
    elif sistolik > 140 and diastolik < 90:
        return 'Hipertensi Sistolik Terisolasi'
    else:
        return 'Tidak Diketahui'

data['Hipertensi_Kategori'] = data.apply(hipertensi_kategori, axis=1)

print("\nData setelah kolom diubah:")
print(data.head())

output_file = 'BMI_Health_Screening_Data_Updated.csv'
data.to_csv(output_file, index=False)
print(f"Data dengan kolom baru telah disimpan ke {output_file}")
```

```python
import pandas as pd
import numpy as np

# Load the existing dataset
data_path = 'Realistic_Health_Screening_Data.csv'
data = pd.read_csv(data_path)

def generate_nadi(is_tachycardia):
    if is_tachycardia:
        return np.random.randint(101, 121)  
    elif np.random.rand() < 0.95:  
        return np.random.randint(60, 101)  
    else:
        return np.random.randint(40, 60)  

def generate_frekuensi_napas():
    if np.random.rand() < 0.95:  
        return np.random.randint(12, 21) 
    else:
        return np.random.choice([
            np.random.randint(21, 30),  
            np.random.randint(8, 12)   
        ])

nadi_values = []
frekuensi_napas_values = []
for _ in range(len(data)):
    frekuensi_napas = generate_frekuensi_napas()
    frekuensi_napas_values.append(frekuensi_napas)

    if frekuensi_napas > 20:  
        nadi_values.append(generate_nadi(is_tachycardia=True))
    elif frekuensi_napas < 12: 
        nadi_values.append(generate_nadi(is_tachycardia=False))
    else:  
        nadi_values.append(generate_nadi(is_tachycardia=False))

data['Nadi (kali/menit)'] = nadi_values
data['Frekuensi Napas (kali/menit)'] = frekuensi_napas_values



output_path = 'nafas_Health_Screening_Data.csv'
data.to_csv(output_path, index=False)
print(f"Data has been updated and saved to {output_path}")
```

```python
import numpy as np

def generate_normal_data(size, mean, std, lower=None, upper=None, decimals=1, frequency=0.99):
    normal_size = int(size * frequency)
    outlier_size = size - normal_size

    normal_data = np.random.normal(mean, std, normal_size)
    if lower is not None:
        normal_data = np.maximum(normal_data, lower)
    if upper is not None:
        normal_data = np.minimum(normal_data, upper)

    if outlier_size > 0:
        outlier_mean = mean * 1.5 if mean > 0 else mean * 0.5
        outlier_std = std * 2
        outlier_data = np.random.normal(outlier_mean, outlier_std, outlier_size)
        if lower is not None:
            outlier_data = np.maximum(outlier_data, lower)
        if upper is not None:
            outlier_data = np.minimum(outlier_data, upper)
    else:
        outlier_data = np.array([])

    data = np.concatenate([normal_data, outlier_data])
    np.random.shuffle(data)
    return np.round(data, decimals)

reduksi_type_probs = [0.94, 0.03, 0.03]  
positive_negative_probs = [0.95, 0.05]  

np.random.seed(42)  
n = len(data)

columns = {
   "hemoglobin": generate_normal_data(n, 14.0, 1.5, lower=10.1, upper=18.0, frequency=0.95),
    "leukosit": generate_normal_data(n, 7500, 1500, lower=4000, upper=11000, decimals=0, frequency=0.95),
    "eritrosit": generate_normal_data(n, 4.75, 0.25, lower=4.0, upper=5.5, frequency=0.95),
    "LED": generate_normal_data(n, 10, 5, lower=0, upper=20, frequency=0.95),
    "eosinofil": generate_normal_data(n, 0.5, 0.3, lower=0, upper=1, frequency=0.95),
    "basofil": generate_normal_data(n, 0.5, 0.3, lower=0, upper=1, frequency=0.95),
    "neutrofil": generate_normal_data(n, 60, 10, lower=50, upper=70, frequency=0.95),
    "lymphosit": generate_normal_data(n, 30, 5, lower=25, upper=33, frequency=0.95),
    "monosit": generate_normal_data(n, 5.5, 1.5, lower=3, upper=8, frequency=0.95),
    "trombosit": generate_normal_data(n, 275000, 50000, lower=150000, upper=400000, decimals=0, frequency=0.95),
    "hematokrit": generate_normal_data(n, 42, 5, lower=37, upper=50, frequency=0.95),
    "MCV": generate_normal_data(n, 87, 5, lower=82, upper=92, frequency=0.95),
    "S G O T": generate_normal_data(n, 25, 6, lower=10, upper=37, frequency=0.95),
    "S G P T": generate_normal_data(n, 30, 8, lower=10, upper=42, frequency=0.95),
    "B U N": generate_normal_data(n, 14, 4, lower=5, upper=23.5, frequency=0.95),
    "kreatinin": generate_normal_data(n, 0.9, 0.3, lower=0.7, upper=1.4, frequency=0.95),
    "asam urat": generate_normal_data(n, 5.0, 1.5, lower=2.6, upper=7.2, frequency=0.95),
    "kolestrol total": generate_normal_data(n, 180, 20, lower=100, upper=200, frequency=0.95),
    "trigliserida": generate_normal_data(n, 150, 30, lower=50, upper=200, frequency=0.95),
    "kolestrol HDL (direct)": generate_normal_data(n, 50, 10, lower=35, upper=70, frequency=0.95),
    "kolestrol LDL (direct)": generate_normal_data(n, 100, 15, lower=50, upper=115, frequency=0.95),
    # "anti HBs": generate_normal_data(n, 500, 300, lower=0, upper=1000, frequency=0.98),
    "HBs Ag Kuantitatif": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "pH pada urine": generate_normal_data(n, 6.5, 0.5, lower=5.0, upper=8.0, frequency=0.95),
    "nitrite pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "protein pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "reduksi pada urine": np.random.choice(["negatif", "positif 1", "positif 2"], n, p=reduksi_type_probs),
    "ketone pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "urobilinogen pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "billirubin pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "silinder pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "kristal pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "yeast pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
    "bakteri pada urine": np.random.choice(["negative", "positif"], n, p=positive_negative_probs),
}

data = data.assign(**columns)

data.info()
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("updated_health_screening_data.csv")  

def random_berat_jenis():
    
    if np.random.rand() < 0.95:  
        return round(np.random.uniform(1.005, 1.030), 3) 
    else:  
        if np.random.rand() < 0.5:  
            return round(np.random.uniform(1.001, 1.004), 3)
        else:  
            return round(np.random.uniform(1.031, 1.040), 3)

def random_eritrosit():
    if np.random.rand() < 0.95:  
        return int(np.random.uniform(0, 3))  
    else:  
        if np.random.rand() < 0.5:  
            return int(np.random.uniform(-1, 0))  
        else:  
            return int(np.random.uniform(3, 6))  

def random_lekosit():
    if np.random.rand() < 0.95:  
        return int(np.random.uniform(0, 6)) 
    else: 
        if np.random.rand() < 0.5: 
            return int(np.random.uniform(-1, 0))  
        else:  
            return int(np.random.uniform(6, 10)) 
def random_gula_darah_puasa():
    if np.random.rand() < 0.95:  
        return int(np.random.uniform(70, 126))  
    else: 
        if np.random.rand() < 0.5:  
            return int(np.random.uniform(126, 200))  
        else:  
            return int(np.random.uniform(200, 400))  

def random_gula_darah_2jam_pp():
    if np.random.rand() < 0.95: 
        return int(np.random.uniform(70, 140)) 
    else: 
        if np.random.rand() < 0.5: 
            return int(np.random.uniform(140, 200)) 
        else:  
            return int(np.random.uniform(200, 400))  
        

df["berat jenis pada urine"] = [random_berat_jenis() for _ in range(len(df))]
df["eritrosit pada urine"] = [random_eritrosit() for _ in range(len(df))]
df["lekosit pada urine"] = [random_lekosit() for _ in range(len(df))]
df["gula darah puasa"] = [random_gula_darah_puasa() for _ in range(len(df))]
df["gula darah 2 jam pp"] = [random_gula_darah_2jam_pp() for _ in range(len(df))]

print(df.head(10))  

df.to_csv("dataset_berat_jenis_updated.csv", index=False)
```
