```python
import pandas as pd
import numpy as np


file_path = "BMI_Health_Screening_Data.csv"


df = pd.read_csv(file_path)


jumlah_tambahan = 1000000


data_baru = pd.DataFrame({
    "id": range(df["id"].max() + 1, df["id"].max() + 1 + jumlah_tambahan),
    "gender": np.random.choice([1, 2], size=jumlah_tambahan),  
})

\
df_baru = pd.concat([df, data_baru], ignore_index=True)

new_file_path = "BMI_Health_Screening_Data_Expanded.csv"
df_baru.to_csv(new_file_path, index=False)

print(f"Dataset baru dengan {jumlah_tambahan} data tambahan telah disimpan di: {new_file_path}")
```
